## Visualizations

```python
print(data["Rate"].dtype)
```

### Clean and convert restaurant ratings

```python
data['Rate_num']=pd.to_numeric(
data['Rate']
    .astype(str)
    .str.replace('out of 5',"")
    .str.strip(),errors="coerce"
)
```

```python
data
```

### Clean and convert approximate cost

```python
data['cost_num']=pd.to_numeric(
data['Approx_cost(for two people)']
    .astype(str)
    .str.replace(',','')
    .str.strip(),errors='coerce'
)
```

```python
data
```

```python
import matplotlib.pyplot as plt
```

## Visualization 1: Distribution of restaurant ratings
### Business question: What ratings are most common among restaurant records?

```python
plt.figure(figsize=(8,5))

plt.hist(data['Rate_num'].dropna(),bins=10,edgecolor='black')

plt.title("Distribution of Resturant ratings")
plt.xlabel("Rating out of 5")
plt.ylabel("Number of records")

plt.tight_layout()
plt.show()
```

## Visualization 2: Online ordering availability
### Business question: How many restaurant records offer online ordering?

```python
order_counts = data["Online_order"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

plt.bar(
    ["No", "Yes"],
    [order_counts.get(0, 0), order_counts.get(1, 0)]
)

plt.title("Online Ordering Availability")
plt.xlabel("Online Ordering")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

## Visualization 3: Distribution of approximate cost
### Business question: What is the typical approximate cost for two people?

```python
plt.figure(figsize=(8,5))

plt.hist(data["cost_num"].dropna(),bins=20,edgecolor="black")

plt.title("Distribution of resturant cost for two")
plt.xlabel("Approximate cost")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

## Visualization 4: Restaurant ratings by online ordering
### Business question: Do ratings differ between records with and without online ordering?

```python
no_order = data.loc[
    data["Online_order"] == 0, "Rate_num"
].dropna()

yes_order = data.loc[
    data["Online_order"] == 1, "Rate_num"
].dropna()

plt.figure(figsize=(7, 5))

plt.boxplot(
    [no_order, yes_order],
    tick_labels=["No Online Order", "Online Order"]
)

plt.title("Ratings by Online Ordering Availability")
plt.xlabel("Online Ordering")
plt.ylabel("Rating out of 5")

plt.tight_layout()
plt.show()
```

## Visualization 5: Most frequently mentioned liked dishes
### Business question: Which dishes are mentioned most frequently?

```python
dishes = (
    data["Dish_liked"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

top_dishes = dishes.value_counts().head(10).sort_values()

plt.figure(figsize=(9, 6))

plt.barh(top_dishes.index, top_dishes.values)

plt.title("Top 10 Most Frequently Mentioned Liked Dishes")
plt.xlabel("Number of Mentions")
plt.ylabel("Dish")

plt.tight_layout()
plt.show()
```

## Visualization 6: Cost versus restaurant rating
### Business question: Are higher-cost restaurant records associated with higher ratings?

```python
plot_data = data.dropna(subset=["Rate_num", "cost_num"])

plt.figure(figsize=(8, 5))

plt.scatter(
    plot_data["cost_num"],
    plot_data["Rate_num"],
    alpha=0.4
)

plt.title("Restaurant Cost vs Rating")
plt.xlabel("Approximate Cost for Two")
plt.ylabel("Rating out of 5")

plt.tight_layout()
plt.show()
```

## Conclusion: Key Findings

This exploratory data analysis examined **8,250 restaurant records** using Python, Pandas, and Matplotlib. After cleaning and standardizing the selected fields, I explored restaurant ratings, approximate costs, online-order availability, table booking, and frequently mentioned liked dishes.

### Key findings

- **Restaurant ratings:** The average rating was approximately **3.87/5**, and the median was **3.9/5**. Numeric rating analysis included 8,190 records; 60 non-numeric ratings could not be converted and were excluded from rating calculations.
- **Online ordering:** **5,697 records (69.1%)** offered online ordering, while 2,553 (30.9%) did not.
- **Restaurant pricing:** The median approximate cost for two people was **₹550**, compared with an average of about **₹704**. The higher mean suggests that some more expensive restaurants pull the average upward.
- **Online ordering and cost:** The median approximate cost was **₹500** among records offering online ordering and **₹700** among records without it. This is an observed association, not proof that online ordering causes lower prices.
- **Table booking:** **1,777 records (21.5%)** offered table booking. Their median rating was **4.2/5**, compared with **3.9/5** for records without table booking. Booking-enabled restaurants also had a higher median cost for two (**₹1,200 versus ₹500**), so price level and restaurant type may be relevant factors.
- **Frequently mentioned liked dishes:** Pasta, burgers, pizza, biryani, and cocktails were among the most frequently mentioned items in the `Dish_liked` field. These are mentions in the available dataset, not necessarily a complete measure of customer demand.

### What I learned

This project strengthened my skills in data inspection, selecting relevant columns, handling duplicates and missing values, cleaning inconsistent text, converting columns to numeric formats, using Pandas for aggregation, and creating histograms, bar charts, boxplots, and scatterplots with Matplotlib.

### Final takeaway and limitations

The analysis shows that online ordering was common in this dataset, restaurant prices varied widely, and table-booking availability was associated with higher median ratings and costs. These patterns can help frame further business questions, but they do not establish causation. The findings are limited to the records analyzed and to the fields available in this dataset.
